```python
X=10
Y=85
D=30

def solution(X,Y,D): # X is the start point, Y is the end point, Z is distance frog jumps
# if the frog is already at the point or if it has already pasted the point then no jump is needed 
    if X >= Y:
        return 0
    else:
    # this will measure the distance the frog needs to jump
        distance = Y - X
        if distance % D == 0: # sees if the value is evenly divided by the jump length
            return distance// D # returns the number of jumps if it was divided evenly  
        else: 
            return(distance//D) +1 # if the value does not divide evenly and has a remainder it will add one extra jump
print(solution(X,Y,D))
```

```python
A = (1,2,3,4,5)
def solution (A):
    N = len(A) #find the numbers of elements in the array 
    for i in range (1, N+1): # will go through each number in the array
         if i not in A: # if a required number is missing, it is not a permutation
            return 0
    return 1 #if all the numbers from 1 to N are present it is permutation 
print(solution(A))
```
